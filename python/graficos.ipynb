```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

from pathlib import Path

sns.set_theme(style="whitegrid")

plt.rcParams["figure.figsize"] = (11, 6)
plt.rcParams["figure.dpi"] = 120

DATA = Path("TABLAS SQL")
OUTPUT = Path("graficas")

OUTPUT.mkdir(exist_ok=True)

resultados = {}

for archivo in DATA.glob("*.csv"):
    resultados[archivo.stem] = pd.read_csv(archivo)

print(f"Archivos cargados: {len(resultados)}")
```

```python
def guardar_grafica(nombre):
    plt.tight_layout()
    plt.savefig(OUTPUT / f"{nombre}.png", bbox_inches="tight")
    plt.show()
    plt.close()
```

## GRÁFICA 1: Clientes recurrentes frente a clientes únicos
Pregunta: ¿Tenemos más clientes que repiten o clientes que solo reservan una vez?

```python
df = resultados["resultado3"].copy()

plt.figure(figsize=(8, 5))

ax = sns.barplot(
    data=df,
    x="customer_type",
    y="customers"
)

plt.title("Clientes recurrentes frente a clientes únicos",
          fontsize=15, fontweight="bold")
plt.xlabel("Tipo de cliente")
plt.ylabel("Número de clientes")

for container in ax.containers:
    ax.bar_label(container, padding=3)

guardar_grafica("01_clientes_recurrentes")
```

## GRÁFICA 2: Los clientes que más gastan
Pregunta: ¿Qué clientes generan un mayor importe acumulado en reservas?

```python
df = resultados["resultado4"].copy()

df = df.sort_values("total_spent", ascending=False).head(10)

plt.figure(figsize=(10, 6))

sns.barplot(
    data=df,
    x="total_spent",
    y="full_name"
)

plt.title("Los 10 clientes que más han gastado",
          fontsize=15, fontweight="bold")
plt.xlabel("Gasto acumulado (€)")
plt.ylabel("Cliente")

guardar_grafica("02_top_clientes")
```

## GRÁFICA 3: Reservas por país
Pregunta: ¿De qué países proceden los clientes que más reservan?

```python
df = resultados["resultado5"].copy()

df = df.sort_values("total_reservations", ascending=False).head(10)

plt.figure(figsize=(10, 6))

sns.barplot(
    data=df,
    x="total_reservations",
    y="country"
)

plt.title("Top 10 países por número de reservas",
          fontsize=15, fontweight="bold")
plt.xlabel("Número de reservas")
plt.ylabel("País")

guardar_grafica("03_reservas_por_pais")
```

## GRÁFICA 4: Los destinos más reservados
Pregunta: ¿Qué destinos tienen mayor demanda?

```python
df = resultados["resultado7"].copy()

df = df.sort_values("total_reservations", ascending=False).head(10)

plt.figure(figsize=(10, 6))

sns.barplot(
    data=df,
    x="total_reservations",
    y="destination"
)

plt.title("Los 10 destinos más reservados",
          fontsize=15, fontweight="bold")
plt.xlabel("Número de reservas")
plt.ylabel("Destino")

guardar_grafica("04_destinos_mas_reservados")
```

## GRÁFICA 5: Popularidad frente a precio de los destinos
Pregunta: ¿Los destinos más populares son también los más caros?

```python
df = resultados["resultado8"].copy()

plt.figure(figsize=(11, 7))

sns.scatterplot(
    data=df,
    x="average_price",
    y="total_reservations",
    size="total_reservations",
    sizes=(50, 300),
    legend=False
)

for _, row in df.iterrows():
    plt.annotate(
        row["destination"],
        (row["average_price"], row["total_reservations"]),
        fontsize=8,
        xytext=(5, 5),
        textcoords="offset points"
    )

plt.title("Popularidad frente a precio de los destinos",
          fontsize=15, fontweight="bold")
plt.xlabel("Precio medio de los tours (€)")
plt.ylabel("Número de reservas")

guardar_grafica("05_popularidad_vs_precio")
```

## GRÁFICA 6: Duración del viaje y precio por día
Pregunta: ¿Cuánto cuesta, de media, cada día de viaje según su duración?

```python
df = resultados["resultado9"].copy()

df = df.sort_values("duration_days")

plt.figure(figsize=(11, 6))

sns.lineplot(
    data=df,
    x="duration_days",
    y="average_price_per_day",
    marker="o",
    linewidth=2
)

plt.title("Precio medio por día según la duración del viaje",
          fontsize=15, fontweight="bold")
plt.xlabel("Duración del viaje (días)")
plt.ylabel("Precio medio por día (€)")
plt.xticks(df["duration_days"])

guardar_grafica("06_precio_por_duracion")
```

## GRÁFICA 7: Estacionalidad de las reservas
Pregunta: ¿En qué meses se producen más reservas?

```python
df = resultados["resultado11"].copy()

meses = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

df["month_name"] = pd.Categorical(
    df["month_name"],
    categories=meses,
    ordered=True
)

plt.figure(figsize=(12, 6))

sns.lineplot(
    data=df,
    x="month_name",
    y="total_reservations",
    hue="booking_year",
    marker="o"
)

plt.title("Evolución mensual de las reservas",
          fontsize=15, fontweight="bold")
plt.xlabel("Mes")
plt.ylabel("Número de reservas")
plt.xticks(rotation=45)

guardar_grafica("07_estacionalidad")
```

## GRÁFICA 8: Importe medio según el canal de reserva
Pregunta: ¿Por qué canal se registra un mayor importe medio por reserva?

```python
df = resultados["resultado12"].copy()

plt.figure(figsize=(9, 6))

ax = sns.barplot(
    data=df,
    x="booking_source",
    y="average_booking_amount"
)

plt.title("Importe medio según el canal de reserva",
          fontsize=15, fontweight="bold")
plt.xlabel("Canal de reserva")
plt.ylabel("Importe medio (€)")

for container in ax.containers:
    ax.bar_label(container, fmt="%.0f €", padding=3)

guardar_grafica("08_importe_por_canal")
```

## GRÁFICA 9: Distribución de los estados de pago
Pregunta: ¿Qué porcentaje de las reservas están completadas, pendientes o reembolsadas?

```python
df = resultados["resultado13"].copy()

plt.figure(figsize=(8, 8))

plt.pie(
    df["total_bookings"],
    labels=df["payment_status"],
    autopct="%1.1f%%",
    startangle=90,
    wedgeprops={"width": 0.45}
)

plt.title("Distribución de los estados de pago",
          fontsize=15, fontweight="bold")

guardar_grafica("09_estados_pago")
```

## GRÁFICA 10: Tours con más reservas completadas
Pregunta: ¿Qué viajes generan más reservas completadas y qué importe acumulan?

```python
df = resultados["resultado15"].copy()

df = df.sort_values(
    "completed_reservations",
    ascending=False
).head(10)

plt.figure(figsize=(11, 7))

sns.barplot(
    data=df,
    x="completed_reservations",
    y="tour_name"
)

plt.title("Top 10 tours por reservas completadas",
          fontsize=15, fontweight="bold")
plt.xlabel("Reservas completadas")
plt.ylabel("Tour")

guardar_grafica("10_tours_completados")
```

## GRÁFICA 11: Contratación de seguros
Pregunta: ¿Qué porcentaje de las reservas tiene un seguro asociado?

```python
df = resultados["resultado16"].iloc[0]

con_seguro = df["bookings_with_insurance"]
sin_seguro = df["total_bookings"] - con_seguro

plt.figure(figsize=(8, 8))

plt.pie(
    [con_seguro, sin_seguro],
    labels=["Con seguro", "Sin seguro"],
    autopct="%1.1f%%",
    startangle=90,
    wedgeprops={"width": 0.45}
)

plt.title("Contratación de seguros en las reservas",
          fontsize=15, fontweight="bold")

guardar_grafica("11_contratacion_seguros")
```

## GRÁFICA 12: Seguros según el importe de la reserva
Pregunta: ¿Los clientes que realizan reservas de mayor importe contratan más seguros?

```python
df = resultados["resultado18"].copy()

plt.figure(figsize=(9, 6))

sns.barplot(
    data=df,
    x="amount_quartile",
    y="insurance_rate"
)

plt.title("Contratación de seguros según el importe",
          fontsize=15, fontweight="bold")
plt.xlabel("Cuartil de importe de reserva (1 = menor importe)")
plt.ylabel("Reservas con seguro (%)")

guardar_grafica("12_seguros_por_importe")
```

## GRÁFICA 13: Rutas de vuelos más frecuentes
Pregunta: ¿Qué rutas aparecen con mayor frecuencia en las reservas?

```python
df = resultados["resultado19"].copy()

df["route"] = (
    df["departure_airport"] + " → " +
    df["arrival_airport"]
)

df = df.sort_values("total_flights", ascending=False).head(10)

plt.figure(figsize=(11, 7))

sns.barplot(
    data=df,
    x="total_flights",
    y="route"
)

plt.title("Las 10 rutas de vuelo más frecuentes",
          fontsize=15, fontweight="bold")
plt.xlabel("Número de vuelos")
plt.ylabel("Ruta")

guardar_grafica("13_rutas_vuelos")
```

## GRÁFICA 14: Ranking final: los viajes que recomendamos
Pregunta: ¿Qué viajes obtienen la mejor puntuación según nuestro modelo de recomendación?

```python
df = resultados["resultado22"].copy()

df = df.sort_values(
    "recommendation_score",
    ascending=False
).head(10)

df["tour_label"] = (
    df["tour_name"] + " - " + df["destination"]
)

plt.figure(figsize=(12, 7))

ax = sns.barplot(
    data=df,
    x="recommendation_score",
    y="tour_label"
)

plt.title("TOP 10: Los viajes que recomendamos",
          fontsize=16, fontweight="bold")
plt.xlabel("Puntuación de recomendación")
plt.ylabel("Viaje")

plt.xlim(0, 100)

for container in ax.containers:
    ax.bar_label(container, fmt="%.1f", padding=3)

guardar_grafica("14_ranking_final")
```
